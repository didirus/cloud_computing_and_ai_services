# Opdrachten voor Linux CLI Opdrachten (Command Line Interface)

## Introductie: Toegang tot je VM via SSH

Om deze opdrachten uit te voeren, moet je inloggen op je Google Cloud VM via SSH. Volg deze stappen:

1. **Navigeer naar Compute Engine:**
   - Klik in het hamburgermenu (☰) linksboven
   - Selecteer **"Compute Engine"** > **"VM instances"**
1. **Vind de gedeelde VM instance** in de lijst
1. **Open SSH verbinding:**
   - Klik op de **"SSH"** knop naast je VM instance naam
   - Er opent een nieuw browservenster met een zwarte terminal
1. **Wacht tot je bent ingelogd** - je ziet een prompt zoals `jouwgebruikersnaam@instance-naam:~$`

Je bent nu klaar om de onderstaande opdrachten uit te voeren!
